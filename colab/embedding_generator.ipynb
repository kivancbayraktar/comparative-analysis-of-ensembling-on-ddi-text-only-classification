```python
import importlib.metadata

def get_dependencies():
    """Return a dict of package -> {version, dependencies[]}"""
    packages = {}
    for dist in importlib.metadata.distributions():
        name = dist.metadata['Name']
        version = dist.version
        deps = [req for req in dist.requires or []]
        packages[name.lower()] = {"version": version, "dependencies": deps}
    return packages


def compare_dependencies(env1, env2):
    """Compare two dependency dictionaries."""
    all_packages = set(env1.keys()) | set(env2.keys())
    differences = {"added": [], "removed": [], "changed": []}

    for pkg in sorted(all_packages):
        if pkg not in env1:
            differences["added"].append((pkg, env2[pkg]["version"]))
        elif pkg not in env2:
            differences["removed"].append((pkg, env1[pkg]["version"]))
        else:
            v1, v2 = env1[pkg]["version"], env2[pkg]["version"]
            if v1 != v2:
                differences["changed"].append((pkg, v1, v2))

    return differences


def print_report(differences, label1="Config A", label2="Config B"):
    print(f"\n📦 Comparison Report ({label1} → {label2})")
    print("-" * 50)

    print("\n➕ Added Packages:")
    for pkg, v in differences["added"]:
        print(f"  {pkg}=={v}")

    print("\n➖ Removed Packages:")
    for pkg, v in differences["removed"]:
        print(f"  {pkg}=={v}")

    print("\n🔄 Version Changes:")
    for pkg, v1, v2 in differences["changed"]:
        print(f"  {pkg}: {v1} → {v2}")

env1 = get_dependencies()  # current environment
```

```python
import pandas as pd
import itertools

def generate_combinations(df, columns):
    """
    Generate new columns in the DataFrame by combining the specified columns in all possible ways.

    Parameters:
    df (pd.DataFrame): The DataFrame containing the columns to combine.
    columns (list): A list of column names to combine.

    Returns:
    pd.DataFrame: DataFrame with the new combination columns added.
    """

    # Generate all possible non-empty combinations of the specified columns
    combinations = []
    for r in range(2, len(columns) + 1):  # Start from 2 to avoid single column combinations
        combinations.extend(itertools.combinations(columns, r))

    # Create new columns by concatenating the values of each combination
    for comb in combinations:
        new_col_name = '_'.join(comb)
        df[new_col_name] = df[list(comb)].apply(lambda row: ' '.join(row.dropna()), axis=1)

    return df


def merge_given_columns(df, columns, new_column_name = "merged_text"):
  df[new_column_name] = df[list(columns)].apply(lambda row: ' '.join(row.dropna()), axis=1)

  return df
```

```python
import os
import json
from typing import Callable, List, Optional, Dict, Any, Type

def generate_embeddings_for_json_object(
    obj_json: dict,
    vector_store_manager_type: Type[BaseVectorStoreManager],
    persist_root: str = "./embeddings",
    new_model_names: Optional[List] = None,
    docs=None
):
    """
    Generate embeddings for all collections in the given JSON object, storing them in a container folder.

    Args:
        obj_json: JSON object with 'id', 'name', and 'collections' keys.
        vector_store_manager_type: The vector store manager class to use.
        persist_root: Root directory for all embeddings.
        new_model_names: Optional list of model names to filter collections.
        docs: Documents to embed (if needed).
    """
    obj_id = obj_json.get("id")
    obj_name = obj_json.get("name")
    collections = obj_json.get("collections", [])

    if not obj_id:
        raise ValueError("JSON object must have an 'id' field.")
    if not collections:
        raise ValueError("No collections found in the given JSON object.")

    # Create container directory for this object
    container_dir = os.path.join(persist_root, str(obj_id))
    os.makedirs(container_dir, exist_ok=True)

    # Call your existing function
    generate_embeddings(
        docs=docs,
        vector_store_manager_type=vector_store_manager_type,
        config_file=None,
        new_model_names=new_model_names,
        collections=collections,
        persist_directory=container_dir
    )
```
