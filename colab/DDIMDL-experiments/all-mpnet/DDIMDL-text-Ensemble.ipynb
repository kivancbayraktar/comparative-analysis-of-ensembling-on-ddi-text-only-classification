```python

print("experiment_name\tcolumn\telapsed_time\taccuracy\tAUC\tAUPR\tf-score\tprecision\trecall")
# for exp_key, results in results_dict.items():
for results in results_list:
    for key in results.keys():
        metric_dict = results[key].metric_dict
        for m in metric_dict:
            accuracy = metric_dict[m].accuracy
            elapsed_time = metric_dict[m].time
            roc_auc = metric_dict[m].roc_auc['macro']
            roc_aupr = metric_dict[m].roc_aupr['micro']
            macro_f_score = metric_dict[m].f1_score['macro']
            macro_precision = metric_dict[m].precision['macro']
            macro_recall = metric_dict[m].recall['macro']
            # print(f"{exp_key}\t{m}\t{accuracy:.4f}\t{macro_f_score:.4f}\t{macro_precision:.4f}\t{macro_recall:.4f}")
            print(f"{key}\t{m}\t{elapsed_time:.4f}\t{accuracy:.4f}\t{roc_auc:.4f}\t{roc_aupr:.4f}\t{macro_f_score:.4f}\t{macro_precision:.4f}\t{macro_recall:.4f}")
```
